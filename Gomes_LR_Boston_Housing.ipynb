#Luke Gomes Multiple Linear Regression - Boston Housing

```python
#import libraries
import pandas as pd
import matplotlib.pyplot as plt

from sklearn import datasets, model_selection
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.linear_model import LinearRegression
```

**Load Data**

```python
df = pd.read_csv("BostonHousing.csv")
df.head()
```

Rename some variables and use it for my new dataframe

```python
df = df.rename(columns={'medv':'TARGET','rm':'RM','lstat': 'LSTAT'})
```

```python
df[['LSTAT','RM','TARGET']].head()
```

```python
plt.figure(figsize=(8,4))

features = df[['LSTAT','RM']]
target = df['TARGET']

plt.subplot(1,2,1)
plt.scatter(df['LSTAT'], target)
plt.xlabel('LSTAT')
plt.ylabel('target')

plt.subplot(1,2,2)
plt.scatter(df['RM'], target)
plt.xlabel('RM')
plt.ylabel('target')

plt.show()
```

**Split data into training and testing data**

```python
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size = 0.2, random_state = 23)
```

**Import Algorithm**

```python
model = LinearRegression()
```

**Fit your model, test on the testing data, and create a visualization if applicable**

```python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
intercept = model.intercept_
coef = model.coef_
```

```python
intercept
```

```python
coef
```

```python
#create visualization
```

```python
plt.figure(figsize=(8,4))
plt.scatter(y_test, y_pred)
plt.plot([0,50,], [0,50], '--k', label = "Correct Prediction")
plt.xlabel('Actual y')
plt.ylabel('Predicted y')
plt.show()
```

**Evaluate Model**


```python
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error :", mse)

mae = mean_absolute_error(y_test, y_pred)
print("Mean Absolute Error:", mae)

r2 = r2_score(y_test, y_pred)
print("R2                 :", r2)
```

**Use Model**

```python
new_houses_data = {
    "LSTAT": [5.0, 4.0, 15.0],
    "RM": [7,6,8]
}

new_houses_df = pd.DataFrame(new_houses_data)
new_houses_df
```

```python
predicted_houses_prices = model.predict(new_houses_df)
predicted_houses_prices
```
